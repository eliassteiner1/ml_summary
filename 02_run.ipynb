```python
import os
import torch
from   torch.utils.data import DataLoader
import numpy as np
import h5py

from P_02_data_loading import MyBatch
from P_02_data_loading import my_collate_fn
from P_02_data_loading import my_jupyter_dataloader
from P_02_data_loading import MyDataset
from P_02_data_loading import MySampler
from P_02_data_loading import MyScheduledBatchSampler

print(f"Cuda Check: {torch.cuda.is_available()}")

with h5py.File("./P_01_data_generation/dataset.h5", "r") as file:
    data = file["data"][...]
    
myset     = MyDataset(data)
mysampler = MyScheduledBatchSampler(myset, 2*[4] + 2*[8], shuffle=True)
myloader  = my_jupyter_dataloader(
    dataset=myset, 
    sampler=mysampler, 
    collate_fn=my_collate_fn, 
    num_workers=0, 
    pin_memory=True
)

onebatch = next(iter(myloader))
onebatch.to("cuda:0")
print(onebatch["label"])


```
