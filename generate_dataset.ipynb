```python
import torch
import torchvision
import torchvision.datasets as datasets
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import h5py

mnist_trainset = datasets.MNIST(root='./data', train=True, download=True, transform=None)
mnist_testset  = datasets.MNIST(root='./data', train=False, download=True, transform=None)


# combine dataset
dataset = mnist_testset # + mnist_trainset

# extract images to np array + labels
samples = []
for el in tqdm(dataset, ncols=120, unit_scale=True):
    img = np.array(el[0])
    lbl = el[1]
    samples.append({"img": img, "lbl": lbl})
    
# STORE: either stacked array (efficient?) or each sample as dict (convenient but inefficient?)

# 1) h5py is perfect for dumping numpy-esque types directly into files!
with h5py.File("testfile.hdf5", "w") as file: 
    dset = file.create_dataset("mydataset", (10, ), dtype="i")
    print(dset[...]) # initialized as empty
    dset[...] = list(range(10))
    print(dset[...])
    dset[...] = np.arange(10)
    print(dset[...])
    
# 2) for complicated samples it might be nice to have one dict per sample with "label", "datax", "datay", "mask", ...
#    but dicts fail for h5yp! 

# 3) so mabye structured array? works!
collection = np.array([
    (1, "alice", 180.5),
    (2, "bob",   176.3),
    (3, "cody",  167.8)
], dtype= [("id", "i4"), ("name", "S10"), ("attr", "f4")])

with h5py.File("testfile.hdf5", "w") as file:
    dset = file.create_dataset("mydataset", data=collection)
    print(dset[...])
    
### so store the mnist files:

```
