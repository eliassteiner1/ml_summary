```python
import numpy as np

# 1D vector operations ---------------------------------------------------------
A = np.random.rand(5, )
B = np.random.rand(5, )

# sum(A)
np.einsum("i -> ", A)

# A*B; elementwise mult
np.einsum("i, i -> i", A, B)

# inner(A, B); dot product (some differences to dot for higher dims...)
np.einsum("i, i -> ", A, B) # same as inner for 1D!

# outer(A, B): outer product
np.einsum("i, j -> ij", A, B)

# 2D array operations ----------------------------------------------------------
C = np.random.rand(5, 5)
D = np.random.rand(5, 5)

# A.T; transpose
np.einsum("ij -> ji", C)

# diag(A); view main diagonal elements
np.einsum("ii -> i", C)

# trace(A); sum main diagonal elements
np.einsum("ii -> ", C)

# sum(A); sum all matrix elements
np.einsum("ij -> ", C)

# sum(A, axis=0); sum along one axis
np.einsum("ij -> j", C) # sum up each column
np.einsum("ij -> i", C) # sum up each row

# A*B; elementwise mult
np.einsum("ij, ij -> ij", C, D)

# A*B.T; elementwise mult with transpose
np.einsum("ij, ji -> ij", C, D) # this kinda just reads in D "in a transpose way"

# dot(A, B); standard matrix mult
np.einsum("ij, jk -> ik", C, D)

# inner(A, B); inner product
np.einsum("ij, kj -> ik", C, D)

# A[:, None] * B; each row of A multiplied by B (elementwise)
np.einsum("ij, kj -> ikj", C, D) # cursed

# A[:, :, None, None] * B; each value of A multiplied by B (elementwise)
np.einsum("ij, kl -> ijkl", C, D) # cursed

# nD arry operations -----------------------------------------------------------
# really useful for general broadcasting and multiple "batch dimensions" while executing multiplications over the last few dimensions. This really highlights how einsum works: first, all the input matrices are broadcast over all the other dimensions that other input arrays have. This kinda results in all input matrices being broadcast / repeated over all "involved" dimensions. Then, these expanded matrices are "overlayed" and all the elements at the same positions are multiplied. Then, all the axes that are not contained in the output string are summed over and thus eliminated. This is especially practical, since it avoids the ubiquitous explicit expanding of matrices to make the following multiplication with broadcasting unambiguous (eg dim(A)=(3, 4) and dim(B)=(N, B, 4, 5) then usually it is necessary to do something like A[None, None, :, :] * B) 

DataA = np.random.rand(20, 16, 3, 4)
DataB = np.random.rand(4, 5)
DataC = np.random.rand(10, 32, 4, 5)

np.einsum("ABij, jk -> ABik", DataA, DataB).shape
np.einsum("ABij, CDjk -> ABCDik", DataA, DataC)

# chaining operations ----------------------------------------------------------
a = np.random.rand(3, )
M = np.random.rand(3, 3)

 # a.T dot A dot a
np.einsum("i, ij, j -> ", a, M, a)

# just a random chain, with batched matrix multiplications. Summing over some axes to get averages or something could easily be done my omitting some signatures after the ->...
np.einsum("ABij, jk, CDlk -> ABCDil", DataA, DataB, DataC)

```
